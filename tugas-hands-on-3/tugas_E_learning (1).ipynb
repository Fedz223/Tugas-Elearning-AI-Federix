```python
# HANDS-ON 3 PANDAS LANJUTAN
# Case 02: Customer Churn Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("Library berhasil diimpor.")
print("Kasus: Customer Churn Analysis")
```

```python
from google.colab import files

uploaded = files.upload()

nama_file = next(iter(uploaded.keys()))
df_mentah = pd.read_csv(nama_file)

print("Nama file:", nama_file)
print("Ukuran dataset:", df_mentah.shape)

display(df_mentah.head())
```

```python

print("Jumlah baris:", len(df_mentah))
print("Jumlah kolom:", len(df_mentah.columns))

print("\nNama kolom:")
print(df_mentah.columns.tolist())

print("\nTipe data dan jumlah nilai terisi:")
df_mentah.info()

print("\nContoh statistik numerik:")
display(df_mentah.describe().T)

print("\nMissing value per kolom:")
display(
    df_mentah.isnull().sum()
    .to_frame("jumlah_missing")
)

print("\nJumlah duplikat baris:",
      df_mentah.duplicated().sum())
```

```python
# Salin dataset asli agar data mentah tetap aman
df_bersih = df_mentah.copy()

# Hitung duplikat sebelum dibersihkan
duplikat_awal = df_bersih.duplicated().sum()

print("Duplikat sebelum dibersihkan:", duplikat_awal)

# Hapus baris duplikat
df_bersih = df_bersih.drop_duplicates().copy()

# Periksa hasil
print("Duplikat setelah dibersihkan:", df_bersih.duplicated().sum())
print("Jumlah baris setelah dibersihkan:", len(df_bersih))
print("Jumlah kolom:", len(df_bersih.columns))
```

```python
missing_awal = df_bersih.isnull().sum()

print("Total missing value:", missing_awal.sum())

display(
    missing_awal[missing_awal > 0]
    .sort_values(ascending=False)
    .to_frame("jumlah_missing")
)
```

```python
kolom_target = "status_pelanggan"

# Hapus baris yang target-nya kosong jika ada
target_kosong = df_bersih[kolom_target].isnull().sum()
if target_kosong > 0:
    df_bersih = df_bersih.dropna(subset=[kolom_target]).copy()

print("Baris dengan target kosong yang dihapus:", target_kosong)

# Tentukan kolom numerik dan kategorikal
kolom_numerik = df_bersih.select_dtypes(include="number").columns.tolist()
kolom_kategorikal = df_bersih.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

# Isi missing value numerik dengan median
for kolom in kolom_numerik:
    if df_bersih[kolom].isnull().any():
        median = df_bersih[kolom].median()
        df_bersih[kolom] = df_bersih[kolom].fillna(median)

# Isi missing value kategorikal dengan modus,
# kecuali kolom target
for kolom in kolom_kategorikal:
    if kolom != kolom_target and df_bersih[kolom].isnull().any():
        modus = df_bersih[kolom].mode(dropna=True)
        if not modus.empty:
            df_bersih[kolom] = df_bersih[kolom].fillna(modus.iloc[0])

print("Total missing value setelah penanganan:",
      df_bersih.isnull().sum().sum())

display(
    df_bersih.isnull().sum()
    .to_frame("missing_setelah")
)
```

```python

kolom_outlier = [
    kolom for kolom in df_bersih.select_dtypes(include="number").columns
    if kolom != "customer_id"
    and kolom != "status_pelanggan"
]

hasil_outlier = []

for kolom in kolom_outlier:
    Q1 = df_bersih[kolom].quantile(0.25)
    Q3 = df_bersih[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    jumlah = (
        (df_bersih[kolom] < batas_bawah) |
        (df_bersih[kolom] > batas_atas)
    ).sum()

    hasil_outlier.append({
        "kolom": kolom,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "batas_bawah": batas_bawah,
        "batas_atas": batas_atas,
        "jumlah_outlier": int(jumlah)
    })

df_outlier = pd.DataFrame(hasil_outlier)

display(
    df_outlier.round(2)
    .sort_values("jumlah_outlier", ascending=False)
)

print("Total temuan outlier per kolom:",
      df_outlier["jumlah_outlier"].sum())
```

```python

# VERIFIKASI LAPORAN KUALITAS

fitur_numerik = [
    "lama_berlangganan_bulan",
    "tagihan_bulanan_idr",
    "total_tagihan_idr",
    "jumlah_layanan_aktif",
    "jumlah_komplain",
    "hari_sejak_interaksi_terakhir"
]

total_outlier_awal = 0
total_outlier_akhir = 0

for kolom in fitur_numerik:
    awal = pd.to_numeric(df_mentah[kolom], errors="coerce")
    akhir = pd.to_numeric(df_bersih[kolom], errors="coerce")

    q1 = awal.quantile(0.25)
    q3 = awal.quantile(0.75)
    iqr = q3 - q1

    bawah = q1 - 1.5 * iqr
    atas = q3 + 1.5 * iqr

    total_outlier_awal += int(
        ((awal < bawah) | (awal > atas)).sum()
    )

    total_outlier_akhir += int(
        ((akhir < bawah) | (akhir > atas)).sum()
    )

print("Outlier pada data mentah:", total_outlier_awal)
print("Outlier pada data bersih, memakai batas awal:", total_outlier_akhir)
print("Missing data bersih:", df_bersih.isnull().sum().sum())
print("Duplikat data bersih:", df_bersih.duplicated().sum())
print("Ukuran data bersih:", df_bersih.shape)
```

```python

# CEK KATEGORI DAN NILAI KOSONG

print("Distribusi jenis kontrak:")
display(df_bersih["jenis_kontrak"].value_counts(dropna=False))

print("Total missing value:", df_bersih.isnull().sum().sum())

print("Kategori status pelanggan:")
display(df_bersih["status_pelanggan"].value_counts(dropna=False))
```

```python

# Simpan jumlah outlier sebelum capping
total_outlier_sebelum = int(df_outlier["jumlah_outlier"].sum())

# Capping berdasarkan batas IQR tiap kolom numerik
for kolom in kolom_outlier:
    Q1 = df_bersih[kolom].quantile(0.25)
    Q3 = df_bersih[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    df_bersih[kolom] = df_bersih[kolom].clip(
        lower=batas_bawah,
        upper=batas_atas
    )

print("Total outlier sebelum capping:", total_outlier_sebelum)
print("Jumlah baris setelah capping:", len(df_bersih))
print("Total missing value:", df_bersih.isnull().sum().sum())

# Audit ulang setelah capping
hasil_setelah = []

for kolom in kolom_outlier:
    Q1 = df_bersih[kolom].quantile(0.25)
    Q3 = df_bersih[kolom].quantile(0.75)
    IQR = Q3 - Q1

    bawah = Q1 - 1.5 * IQR
    atas = Q3 + 1.5 * IQR

    jumlah = (
        (df_bersih[kolom] < bawah) |
        (df_bersih[kolom] > atas)
    ).sum()

    hasil_setelah.append({
        "kolom": kolom,
        "outlier_setelah": int(jumlah)
    })

display(pd.DataFrame(hasil_setelah))
```

```python

# Kolom numerik yang seharusnya berisi angka
kolom_angka = [
    "lama_berlangganan_bulan",
    "tagihan_bulanan_idr",
    "total_tagihan_idr",
    "jumlah_layanan_aktif",
    "jumlah_komplain",
    "hari_sejak_interaksi_terakhir"
]

# Pastikan kolom numerik bertipe angka
for kolom in kolom_angka:
    df_bersih[kolom] = pd.to_numeric(
        df_bersih[kolom], errors="coerce"
    )

# Rapikan spasi dan kapitalisasi kategori
kolom_teks = [
    "jenis_kontrak",
    "layanan_internet",
    "dukungan_teknis",
    "metode_pembayaran",
    "status_pelanggan"
]

for kolom in kolom_teks:
    df_bersih[kolom] = (
        df_bersih[kolom]
        .astype("string")
        .str.strip()
    )

# Normalisasi variasi penulisan jenis kontrak
kontrak_normal = (
    df_bersih["jenis_kontrak"]
    .str.lower()
    .str.replace("@", "a", regex=False)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

peta_kontrak = {
    "bulanan": "Bulanan",
    "tahunan": "Tahunan",
    "dua tahun": "Dua Tahun"
}

df_bersih["jenis_kontrak"] = kontrak_normal.map(peta_kontrak)

# Pastikan ID diperlakukan sebagai identitas, bukan angka analisis
df_bersih["customer_id"] = df_bersih["customer_id"].astype("string")

print("Tipe data setelah perbaikan:")
display(df_bersih.dtypes.to_frame("tipe_data"))

print("\nKategori jenis kontrak:")
display(df_bersih["jenis_kontrak"].value_counts(dropna=False))

print("\nMissing value setelah normalisasi:")
display(df_bersih.isnull().sum().to_frame("jumlah_missing"))
```

```python

fitur_outlier = [
    kolom for kolom in kolom_angka
    if kolom in df_mentah.columns
]

batas_iqr_awal = {}

for kolom in fitur_outlier:
    nilai = pd.to_numeric(df_mentah[kolom], errors="coerce").dropna()
    q1 = nilai.quantile(0.25)
    q3 = nilai.quantile(0.75)
    iqr = q3 - q1
    batas_iqr_awal[kolom] = (
        q1 - 1.5 * iqr,
        q3 + 1.5 * iqr
    )

def hitung_outlier_dengan_batas(df, batas):
    total = 0
    for kolom, (bawah, atas) in batas.items():
        nilai = pd.to_numeric(df[kolom], errors="coerce")
        total += int(((nilai < bawah) | (nilai > atas)).sum())
    return total

laporan_kualitas = pd.DataFrame([
    {
        "metrik": "Jumlah baris",
        "sebelum": len(df_mentah),
        "sesudah": len(df_bersih)
    },
    {
        "metrik": "Jumlah kolom",
        "sebelum": len(df_mentah.columns),
        "sesudah": len(df_bersih.columns)
    },
    {
        "metrik": "Baris duplikat",
        "sebelum": int(df_mentah.duplicated().sum()),
        "sesudah": int(df_bersih.duplicated().sum())
    },
    {
        "metrik": "Total missing value",
        "sebelum": int(df_mentah.isnull().sum().sum()),
        "sesudah": int(df_bersih.isnull().sum().sum())
    },
    {
        "metrik": "Outlier berdasarkan batas IQR awal",
        "sebelum": hitung_outlier_dengan_batas(df_mentah, batas_iqr_awal),
        "sesudah": hitung_outlier_dengan_batas(df_bersih, batas_iqr_awal)
    }
])

display(laporan_kualitas)

print("Interpretasi:")
print("- Baris yang duplikat dihapus jika ditemukan.")
print("- Missing value ditangani menggunakan median/modus.")
print("- Outlier dibatasi menggunakan capping IQR.")
print("- Normalisasi kategori dilakukan agar label konsisten.")
```

```python

kolom_target = "status_pelanggan"

print("Distribusi status pelanggan:")
display(
    df_bersih[kolom_target]
    .value_counts()
    .to_frame("jumlah_pelanggan")
)

print("\nPersentase status pelanggan:")
display(
    (df_bersih[kolom_target].value_counts(normalize=True) * 100)
    .round(2)
    .to_frame("persentase")
)

def analisis_churn(kolom):
    tabel = pd.crosstab(
        df_bersih[kolom],
        df_bersih[kolom_target]
    )

    if "Churn" not in tabel.columns:
        tabel["Churn"] = 0

    tabel["total_pelanggan"] = tabel.sum(axis=1)
    tabel["jumlah_churn"] = tabel["Churn"]
    tabel["persentase_churn"] = (
        tabel["jumlah_churn"] / tabel["total_pelanggan"] * 100
    ).round(2)

    return tabel.sort_values("persentase_churn", ascending=False)

for kolom in [
    "jenis_kontrak",
    "layanan_internet",
    "metode_pembayaran"
]:
    print(f"\nAnalisis berdasarkan {kolom}:")
    display(analisis_churn(kolom))
```

```python

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Mulai dari dataset mentah; jangan gunakan df_bersih yang sudah
# diimputasi atau di-capping menggunakan seluruh dataset.
df_model = df_mentah.copy()

# Hilangkan duplikat dan baris tanpa target
df_model = df_model.drop_duplicates().dropna(
    subset=["status_pelanggan"]
).copy()

# Normalisasi kategori kontrak sebelum split (transformasi deterministik)
kontrak_model = (
    df_model["jenis_kontrak"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.replace("@", "a", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

df_model["jenis_kontrak"] = kontrak_model.map({
    "bulanan": "Bulanan",
    "tahunan": "Tahunan",
    "dua tahun": "Dua Tahun"
})

# Pastikan fitur numerik terbaca sebagai angka
for kolom in kolom_angka:
    df_model[kolom] = pd.to_numeric(
        df_model[kolom], errors="coerce"
    )

# ID tidak digunakan sebagai fitur prediksi
X = df_model.drop(columns=["customer_id", "status_pelanggan"])
y = df_model["status_pelanggan"].astype(str).str.strip()

# Split dahulu, sebelum imputasi, scaling, dan encoding
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

kolom_num_model = X_train.select_dtypes(include="number").columns.tolist()
kolom_cat_model = [
    kolom for kolom in X_train.columns
    if kolom not in kolom_num_model
]

pipeline_numerik = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

pipeline_kategorikal = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numerik", pipeline_numerik, kolom_num_model),
    ("kategorikal", pipeline_kategorikal, kolom_cat_model)
])

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000, random_state=42))
])

# Fit hanya pada data training
model.fit(X_train, y_train)

# Evaluasi pada data testing
prediksi = model.predict(X_test)

print("Jumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))
print("Accuracy:", round(accuracy_score(y_test, prediksi), 4))

print("\nClassification report:")
print(classification_report(y_test, prediksi, zero_division=0))

print("Confusion matrix (urutan label):", model.classes_)
display(pd.DataFrame(
    confusion_matrix(y_test, prediksi, labels=model.classes_),
    index=[f"Aktual {label}" for label in model.classes_],
    columns=[f"Prediksi {label}" for label in model.classes_]
))
```
